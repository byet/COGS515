## 4. Control flow and functions

Blocks start with a colon and are **indented**. `if` chooses between blocks, `for` repeats over the items of a sequence, and `while` repeats as long as a condition holds.

````python
score, absent = 72, False
if score >= 90:
    print("excellent")
elif score >= 60 and not absent:     # and, or, not combine conditions
    print("pass")
else:
    print("fail")
label = "pass" if score >= 60 else "fail"   # a conditional expression: the same choice on one line
print(label, score != 100, "x" not in ["a", "b"])

for i, word in enumerate(["search", "logic", "learning"]):   # position and item
    print(i, word)

for a, b in zip([1, 2, 3], ["x", "y", "z"]):                  # walk two lists together
    print(a, b)

squares = [n ** 2 for n in range(10) if n % 2 == 0]           # list comprehension
print(squares)
print(sum(n ** 2 for n in range(10)))                         # the same, without building a list

n = 1
while n < 50:                    # while repeats as long as its condition holds
    n = n * 2
print(n)

frontier = ["c", "b", "a"]
while frontier:                  # a non-empty list counts as True: this is the shape of every search loop
    item = frontier.pop()        # take the last item out
    if item == "b":
        continue                 # skip the rest of this round
    print("visiting", item)

n = 0
while True:                      # loop "forever"...
    n += 1                       # n = n + 1
    if n ** 2 > 50:
        break                    # ...until break leaves the loop
print(n)
````

## 5. Classes

A **class** is a blueprint for objects that keep their own **state** (attributes) and have **methods** (functions that belong to the object).
- `__init__` runs when an object is created.
- `self` is the object itself. You write it as the first parameter of every method but don't pass it when you call the method.

````python
class Circle:
    def __init__(self, radius=1):
        self.radius = radius         # an attribute: each circle has its own

    def area(self):
        return 3.14159 * self.radius ** 2

    def __str__(self):               # used by print() and str()
        return f"Circle with radius {self.radius}"

    def __eq__(self, other):         # used by ==
        return self.radius == other.radius

c1, c2 = Circle(10), Circle(3)
c2.radius = 100                      # attributes can be changed directly
print(c1.radius, c2.radius, round(c1.area(), 1))
print(c1, c1 == Circle(10), c1 == c2)
````

Methods whose names start and end with two underscores are **special methods**: Python calls them for you.

| You write | Python calls |
|---|---|
| `print(c)` or `str(c)` | `c.__str__()` |
| `c1 == c2` | `c1.__eq__(c2)` |
| `c1 + c2` | `c1.__add__(c2)` |
| `c(x)` | `c.__call__(x)` |

### Worked example: the vacuum world (Week 1)

Two squares, A and B, each clean or dirty. The agent perceives its location and whether that square is dirty. Scoring:
- **+10** for sucking up dirt
- **−1** for each move

Look at `step`: the environment calls `agent.program(percept)`. The agent's behaviour is a function stored in the agent, so you can swap behaviours by passing a different function. HW1 builds on this code.

````python
import random

loc_A, loc_B = 0, 1

class SimpleVacuumEnvironment:
    def __init__(self):
        self.status = {loc_A: random.choice(["Clean", "Dirty"]),
                       loc_B: random.choice(["Clean", "Dirty"])}
        self.agent = None

    def add_agent(self, agent, location=None):
        agent.location = location if location is not None else random.choice([loc_A, loc_B])
        self.agent = agent

    def percept(self, agent):
        return agent.location, self.status[agent.location]

    def step(self):
        action = self.agent.program(self.percept(self.agent))
        self.execute_action(self.agent, action)

    def execute_action(self, agent, action):
        if action == "Right":
            agent.location = loc_B
            agent.performance -= 1
        elif action == "Left":
            agent.location = loc_A
            agent.performance -= 1
        elif action == "Suck":
            if self.status[agent.location] == "Dirty":
                agent.performance += 10
            self.status[agent.location] = "Clean"

    def run(self, steps=100):
        for _ in range(steps):          # _ : a loop variable we never use
            self.step()


class ReflexVacuumAgent:
    def __init__(self, program):
        self.performance = 0
        self.program = program


def reflex_program(percept):
    location, status = percept
    if status == "Dirty":
        return "Suck"
    return "Right" if location == loc_A else "Left"
````

````python
random.seed(1)   # fixes the random choices so every run gives the same result
env = SimpleVacuumEnvironment()
agent = ReflexVacuumAgent(reflex_program)
env.add_agent(agent)
print("start:", env.status, "agent at", agent.location)
env.run(10)
print("after 10 steps:", env.status, "performance", agent.performance)
````

## 6. Tools for search: `deque`, `Counter`, `defaultdict`, `heapq`

- `deque` is a double-ended queue. Use `popleft()` for a first-in, first-out queue (BFS) or `pop()` for a stack (DFS). It is fast at both ends.
- `Counter` counts things.
- `defaultdict` creates missing keys automatically.
- `heapq` turns a list into a priority queue, always giving back the smallest item first. It is the tool for uniform-cost search and A*.

````python
from collections import deque, Counter, defaultdict

q = deque(["a", "b"])
q.append("c")
print(q.popleft(), q.pop(), q)       # 'a' from the front, 'c' from the back

words = "the cat and the dog and the bird".split()
print(Counter(words).most_common(2))

by_len = defaultdict(list)
for w in words:
    by_len[len(w)].append(w)
print(dict(by_len))
````

**Common error:** `my_list.pop(0)` works but gets slow on long lists, because every item has to shift. For queues, use `deque.popleft()`.

````python
import heapq

frontier = []
heapq.heappush(frontier, (5, "far"))
heapq.heappush(frontier, (1, "near"))
heapq.heappush(frontier, (3, "middle"))
print(heapq.nsmallest(2, frontier))   # the 2 smallest, as a sorted list (a sorted list is also a valid heap)
print(heapq.heappop(frontier))        # (1, 'near'): smallest priority first
print(frontier)
````

**Common error: ties.** When two entries have the same priority, `heapq` compares the second items. If those are objects, such as search nodes, it fails:

````python
class Item:
    def __init__(self, name):
        self.name = name

h = []
try:
    heapq.heappush(h, (1, Item("A")))
    heapq.heappush(h, (1, Item("B")))   # same priority, so heapq compares the two Items
except TypeError as e:
    print("TypeError:", e)
````

The fix is to put a running counter between the priority and the object. Ties are then broken by insertion order, and the objects are never compared. After `from itertools import count`, don't use `count` as a variable name, or the import is lost.

````python
from itertools import count

tie = count()
h = []
heapq.heappush(h, (1, next(tie), Item("A")))
heapq.heappush(h, (1, next(tie), Item("B")))
priority, _, item = heapq.heappop(h)
print(priority, item.name)
````

## 7. A little NumPy and matplotlib

A NumPy array applies arithmetic to every element at once. On **log–log axes** a power law, $y = c\,x^{a}$, is a straight line whose slope is the exponent $a$. HW1 uses this to test Zipf's law. To show two series on one figure, give each a `label` and call `plt.legend()`.

````python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([1, 2, 3, 4])
print(a * 10, a.mean(), a.sum(), np.log10(a))
print([1, 2] * 2, np.array([1, 2]) * 2)   # common error: * repeats a list but multiplies an array

counts = {"the": 50, "of": 25, "cat": 5}
freqs = np.array(sorted(counts.values(), reverse=True))   # from word counts to a ranked frequency array
print(freqs)

ranks = np.arange(1, 101)
plt.loglog(ranks, 1000 / ranks, "o", markersize=3, label="exponent -1")          # a perfect power law
plt.loglog(ranks, 1000 / ranks ** 1.5, "s", markersize=3, label="exponent -1.5")
plt.xlabel("rank")
plt.ylabel("frequency")
plt.legend()
plt.show()
````

### 5. Node

```python
class Node:
    def __init__(self, state, parent=None, action=None, path_cost=0):
        self.state = state
        self.parent = parent
        self.action = action
        self.path_cost = path_cost
        self.depth = 0 if parent is None else parent.depth + 1

    def path(self):
        states, node = [], self
        while node is not None:
            states.append(node.state)
            node = node.parent
        return states[::-1]
```
